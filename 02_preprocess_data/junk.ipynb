```python
def resize_recursively(image, scale=1.5, max_size=400):
    """
    Recursively scales an image by a factor of 1.5 until either width or height reaches at least 400 pixels.

    :param image: Input image as a NumPy array (cv2.imread output).
    :param scale: Scaling factor (default: 1.5x per iteration).
    :param max_size: The threshold for width or height to stop scaling (default: 400 pixels).
    :return: Resized image as a NumPy array.
    """
    height, width = image.shape[:2]

    # Base case: stop if either dimension reaches or exceeds max_size
    if width >= max_size or height >= max_size:
        return image

    # Compute new dimensions
    new_width = int(width * scale)
    new_height = int(height * scale)

    # Resize the image
    resized_image = cv2.resize(image, (new_width, new_height), interpolation=cv2.INTER_LINEAR)

    # Recursively resize further if conditions are met
    return resize_recursively(resized_image, scale, max_size)


image_path='/Users/levgolod/Projects/car_classifier/data/autotrader/vehicle_images/make-toyota/model-tacoma/vehicle_id-730197477/2f8da6e615624dcba09c22b1ec1e4479.jpg'
image = cv2.imread(image_path)
image_embigenned = resize_recursively(image)

plt.imshow(image_embigenned)
plt.axis("off")  # Hide axes
plt.show()
```
